# Exploratory Data Analysis

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Create synthetic data to test nbstripout hook
np.random.seed(42)

n = 100

df = pd.DataFrame({
    "patient_id":  range(1, n + 1),
    "age":         np.random.randint(40, 90, n),
    "lvef":        np.random.uniform(20, 65, n),
    "systolic_bp": np.random.randint(90, 180, n),
    "mortality":   np.random.randint(0, 2, n),
})

print(df.shape)
df.head(10)
```

```python
df.describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# LVEF distribution by mortality
for outcome, label, color in zip([0, 1], ["Survived", "Died"], ["steelblue", "tomato"]):
    subset = df[df["mortality"] == outcome]["lvef"]
    axes[0].hist(subset, bins=20, alpha=0.6, label=label, color=color)

axes[0].set_title("LVEF Distribution by Outcome")
axes[0].set_xlabel("LVEF (%)")
axes[0].set_ylabel("Count")
axes[0].legend()

# Age vs Systolic BP
scatter = axes[1].scatter(
    df["age"],
    df["systolic_bp"],
    c=df["mortality"],
    cmap="coolwarm",
    alpha=0.7,
    edgecolors="none"
)
axes[1].set_title("Age vs Systolic BP")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Systolic BP")
plt.colorbar(scatter, ax=axes[1], label="Mortality")

plt.tight_layout()
plt.show()
```
